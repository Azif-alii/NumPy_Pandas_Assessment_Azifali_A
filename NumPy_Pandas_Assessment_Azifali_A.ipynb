#*PART -A (EASY)*

```python

#Q1 Dataset loading anmd inspection
import numpy as np
import pandas as pd

df = pd.read_csv("retail_sales.csv")
#it shows first five records
print(df.head())
#it shows number of columns and rows
print(df.shape)
#it shows number columns
print(df.columns)
#it shows data types
print(df.dtypes)
```

```python
#Q2 Numpy Operations
q = df["Quantity"].to_numpy()

# to show total number of quantity sold
print("Total:", np.sum(q))
# to shows average number of quantity
print("Average:", np.mean(q))
# to shows maximum number of quantity
print("Maximum:", np.max())
# to shows minimum number of quantity
print("Minimum:", np.min(q))
# to shows the transactions where the quantity solder greater than >3
print(df[df["Quantity"] > 3])
```

```python
#Q3 Basic pandas Analysis

# 11. to display all transactions from kochi
print('\n display all transactions from kochi \n',(df[df["City"] == "Kochi"]))
# 12. to display transactions belonging to the electronics category
print('\n display transactions belonging to the electronics category \n',(df[df["Category"] == "Electronics"]))
# 13. to display find all transactions where the unit price is > 10,000
print('\n display all transactions where the unit price is greater than ₹10,000 \n',
(df[df["Unit_Price"] > 10000]))
#14. to display only the product,city, and quantity columns
print('\n display only the Product, City, and Quantity columns \n',
(df[["Product", "City", "Quantity"]]))
#15. to display the sort transactions by unit price and descending order
print('\n to display Sort the transactions by unit price in descending order \n',
(df.sort_values("Unit_Price", ascending=False)))
```

#*PART -B (MEDIUM*)

```python
#Q4 Calculate Revenue

# 16 create gross_sales,quantity,unit_price
df["Gross_Sales"] = df["Quantity"] * df["Unit_Price"]
# 17 create net_sales by deducting the discount from gross sales
df["Net_Sales"] = df["Gross_Sales"] * (1 - df["Discount"])
# 18 to display the updated dataframe
print("Updated DataFrame with Gross_Sales and Net_Sales:")
print(df.head())
# 19 to display claculate total gross sales
total_gross_sales = df["Gross_Sales"].sum()
print(f"\nTotal Gross Sales: ₹{total_gross_sales:,.2f}")
# 20 to display calculate total net sales
total_net_sales = df["Net_Sales"].sum()
print(f"Total Net Sales: ₹{total_net_sales:,.2f}")
# 21 to dispaly transactions with highest net sales
max_net_sales_row = df.loc[df["Net_Sales"].idxmax()]
print("\nTransaction with Highest Net Sales:")
print(max_net_sales_row)
# 22 to display calculate the total discount amount  given by the company
total_discount = (df["Gross_Sales"] * df["Discount"]).sum()
print(f"\nTotal Discount Given by the Company: ₹{total_discount:,.2f}")
```

```python
#Q5 Category-Wise Analysis

category_report = df.groupby("Category").agg(
    #23 Calculate total net sales for each product category
    Total_Net_Sales=("Net_Sales", "sum"),
    #24 Find the total quantity sold in each category
    Total_Quantity=("Quantity", "sum"),
    #26	Calculate the average net sales per transaction for each category
    Average_Net_Sales=("Net_Sales", "mean")
).reset_index()

#25 Identify the category generating the highest net sales
top_category = category_report.loc[category_report["Total_Net_Sales"].idxmax(), "Category"]

print(category_report)
print(f"\nCategory generating highest net sales: {top_category}")
```

```python
#Q6 City-Wise Performance

city_summary = df.groupby("City").agg(
    #27 Calculate total net sales for each city
    Total_Net_Sales=("Net_Sales", "sum"),
    #28 Count the number of transactions in each city
    Transaction_Count=("Order_ID", "count"),
    #29 Identify the city with the highest net sales
    Average_Order_Value=("Net_Sales", "mean")
).reset_index()

#30 Calculate average order value for each city
top_city = city_summary.loc[city_summary["Total_Net_Sales"].idxmax(), "City"]
print("\nQ6 City Summary:\n", city_summary)
print("Top City:", top_city)
```

```python
#Q8 Product performance Report

#33 Generate a product-wise report containing product name, total quantity sold, total net sales, average unit price, and number of transactions
product_report = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Net_Sales=("Net_Sales", "sum"),
    Average_Unit_Price=("Unit_Price", "mean"),
    Number_of_transactions =("Order_ID", "count")
    #34 Sort the report by total net sales in descending order
).reset_index().sort_values(by="Total_Net_Sales", ascending=False)
#35 Identify the top three products based on revenue
top_3_products = product_report.head(3)
print("\n Top 3 Products:\n", top_3_products)
```

```python
#Q10 Business Decision-Making

#42 Generate a report showing product, total quantity sold, total net sales, and average net sales
bus_report = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Net_Sales=("Net_Sales", "sum"),
    Average_Net_Sales=("Net_Sales", "mean")
).reset_index()
# 40 Identify products with total quantity sold below 5 units
low_qty_products = bus_report[bus_report["Total_Quantity"] < 5]["Product"].tolist()
#41 Identify products whose average net sales per transaction are below ₹10,000
low_avg_net_products = bus_report[bus_report["Average_Net_Sales"] < 10000]["Product"].tolist()

print(bus_report)
print(f"\nProducts with Total Quantity Sold below 5 units: {low_qty_products}")
print(f"Products with Average Net Sales below ₹10,000: {low_avg_net_products}")
```
